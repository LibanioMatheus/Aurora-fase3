Notebook responsável pela **organização dos dados** da colônia.

Aplica os conceitos de **Estruturas de Dados** vistos na fase:

- **listas** — séries históricas de sensores;
- **dicionários (tabela hash)** — acesso rápido em chave-valor;
- **hierarquia** — dicionário aninhado com os subsistemas.

Usa apenas Python puro, sem bibliotecas externas.

## 1. Estrutura hierárquica da colônia

A função `criar_colonia()` monta a estrutura completa, organizada
do subsistema mais geral ao mais específico.

```python
def criar_colonia():
    """
    Monta a estrutura hierarquica da colonia.

    Modelo:
        colonia
          |- energetico  -> solar / eolico / reserva
          |- ambiental
          |- operacional -> modulos

    Retorna:
        dict: estrutura completa da colonia.
    """
    colonia = {
        "energetico": {
            "solar": {
                "geracao": 45,
                "historico_geracao": [30, 34, 38, 41, 45],
            },
            "eolico": {
                "geracao": 25,
                "historico_vento": [8, 10, 12, 13, 15],
                "historico_geracao": [18, 21, 25, 27, 31],
            },
            "reserva_baterias": 60,
        },
        "ambiental": {
            "temperatura_interna": 22,
            "temperatura_externa": -55,
            "velocidade_vento": 14,
            "previsao_tempestade": False,
        },
        "operacional": {
            "consumo_total": 70,
            "modulos": {
                "suporte_vida": {"consumo": 30, "essencial": True},
                "estufa": {"consumo": 20, "essencial": False},
                "laboratorio": {"consumo": 12, "essencial": False},
                "comunicacao": {"consumo": 8, "essencial": True},
            },
        },
    }
    return colonia
```

## 2. Navegação entre subsistemas

A função `acessar()` percorre a hierarquia usando uma lista de
chaves como caminho, reproduzindo a navegação entre estruturas
hierárquicas trabalhada na disciplina.

```python
def acessar(colonia, caminho):
    """
    Navega pela hierarquia usando uma lista de chaves.

    Exemplo:
        acessar(colonia, ["energetico", "solar", "geracao"]) -> 45

    Retorna:
        valor encontrado, ou None se o caminho nao existir.
    """
    atual = colonia
    for chave in caminho:
        if isinstance(atual, dict) and chave in atual:
            atual = atual[chave]
        else:
            return None
    return atual
```

## 3. Funções auxiliares de leitura

```python
def energia_total_gerada(colonia):
    """Soma a geracao solar e eolica (acesso via chave-valor)."""
    solar = acessar(colonia, ["energetico", "solar", "geracao"])
    eolico = acessar(colonia, ["energetico", "eolico", "geracao"])
    return solar + eolico


def listar_modulos_nao_essenciais(colonia):
    """Retorna os nomes dos modulos que NAO sao essenciais."""
    modulos = acessar(colonia, ["operacional", "modulos"])
    nao_essenciais = []
    for nome in modulos:
        if not modulos[nome]["essencial"]:
            nao_essenciais.append(nome)
    return nao_essenciais


def resumo(colonia):
    """Imprime um resumo legivel do estado da colonia."""
    print("=" * 50)
    print("RESUMO DO ESTADO DA COLONIA AURORA SIGER")
    print("=" * 50)
    print("Geracao solar .......:", acessar(colonia, ["energetico","solar","geracao"]))
    print("Geracao eolica ......:", acessar(colonia, ["energetico","eolico","geracao"]))
    print("Geracao total .......:", energia_total_gerada(colonia))
    print("Reserva (baterias) ..:", acessar(colonia, ["energetico","reserva_baterias"]))
    print("Consumo total .......:", acessar(colonia, ["operacional","consumo_total"]))
    print("Temp. interna .......:", acessar(colonia, ["ambiental","temperatura_interna"]), "C")
    print("Velocidade do vento .:", acessar(colonia, ["ambiental","velocidade_vento"]))
    print("Tempestade prevista .:", acessar(colonia, ["ambiental","previsao_tempestade"]))
    print("=" * 50)
```

## 4. Demonstração

Execute a célula abaixo para ver a organização dos dados em ação.

```python
colonia = criar_colonia()
resumo(colonia)

print()
print("Acesso direto -> geracao solar:",
      acessar(colonia, ["energetico", "solar", "geracao"]))
print("Modulos nao essenciais:", listar_modulos_nao_essenciais(colonia))
```
