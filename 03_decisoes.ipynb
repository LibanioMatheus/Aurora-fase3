## 1. Análise do uso de energia

Compara geração, consumo e reserva e devolve uma ação simples.

```python
def analisar_energia(geracao, consumo, reserva=0):
    """
    Compara geracao, consumo e reserva e devolve uma acao simples.

    Regras:
        consumo > geracao  -> RISCO
        geracao > consumo  -> SOBRA
        geracao == consumo -> EQUILIBRIO

    Retorna:
        dict com 'situacao' e 'mensagem'.
    """
    saldo = geracao - consumo

    if consumo > geracao:
        situacao = "RISCO"
        mensagem = "ALERTA: consumo maior que geracao"
    elif geracao > consumo:
        situacao = "SOBRA"
        mensagem = "SUGESTAO: armazenar energia excedente"
    else:
        situacao = "EQUILIBRIO"
        mensagem = "OK: geracao e consumo equilibrados"

    return {
        "geracao": geracao,
        "consumo": consumo,
        "reserva": reserva,
        "saldo": saldo,
        "situacao": situacao,
        "mensagem": mensagem,
    }
```

## 2. Regras de decisão (lógica do sistema)

Combina múltiplos critérios com lógica booleana, em ordem de
prioridade:

1. energia < 30 **E** consumo alto → Modo de Emergência
2. tempestade prevista **E** energia < 50 → Modo de Economia
3. energia < 50 → Reduzir Consumo
4. caso contrário → Sistemas Normais

```python
def decidir_acao(energia, consumo, previsao_tempestade=False):
    """
    Toma a decisao operacional principal da colonia.

    'consumo alto' = consumo >= 60.

    Retorna:
        dict com nivel, acao e detalhes da decisao.
    """
    consumo_alto = consumo >= 60
    energia_critica = energia < 30
    energia_baixa = energia < 50

    if energia_critica and consumo_alto:
        nivel = "CRITICO"
        acao = "ATIVAR MODO DE EMERGENCIA"
        detalhe = "Energia critica com consumo alto."
    elif previsao_tempestade and energia_baixa:
        nivel = "ALTO"
        acao = "ATIVAR MODO DE ECONOMIA"
        detalhe = "Tempestade de areia prevista e energia baixa."
    elif energia_baixa:
        nivel = "MEDIO"
        acao = "REDUZIR CONSUMO"
        detalhe = "Energia abaixo do nivel seguro (50)."
    else:
        nivel = "NORMAL"
        acao = "MANTER SISTEMAS NORMAIS"
        detalhe = "Condicoes dentro do esperado."

    return {
        "entrada": {"energia": energia, "consumo": consumo,
                    "tempestade": previsao_tempestade},
        "nivel": nivel,
        "acao": acao,
        "detalhe": detalhe,
    }
```

## 3. Priorização de módulos

Os módulos essenciais (suporte à vida, comunicação) ficam sempre
ligados. Em modo de economia, os não essenciais são desligados.

```python
def priorizar_modulos(modulos, modo_economia):
    """
    Define quais modulos ficam LIGADOS e quais sao DESLIGADOS.

    Retorna:
        dict com listas 'ligados' e 'desligados' e o consumo final.
    """
    ligados = []
    desligados = []
    consumo_final = 0

    for nome in modulos:
        info = modulos[nome]
        if info["essencial"]:
            ligados.append(nome)
            consumo_final += info["consumo"]
        elif modo_economia:
            desligados.append(nome)
        else:
            ligados.append(nome)
            consumo_final += info["consumo"]

    return {
        "ligados": ligados,
        "desligados": desligados,
        "consumo_final": consumo_final,
    }
```

## 4. Demonstração

Testa a lógica de decisão em quatro cenários diferentes.

```python
print("Analise de energia:")
print(" ", analisar_energia(40, 70)["mensagem"])
print(" ", analisar_energia(80, 30)["mensagem"])

print()
print("Decisoes por cenario:")
cenarios = [
    ("Energia critica + consumo alto", 25, 70, False),
    ("Tempestade prevista + energia baixa", 45, 40, True),
    ("Energia baixa, sem tempestade", 48, 35, False),
    ("Condicoes normais", 90, 50, False),
]
for nome, e, c, t in cenarios:
    d = decidir_acao(e, c, t)
    print(f"  - {nome}: [{d['nivel']}] {d['acao']}")
```
